## Question 1

* Whats the distribution of loan amounts?

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='LoanOriginalAmount', bins=30)
plt.title('Distribution of Loan Amounts', fontsize=14, fontweight='bold')
plt.xlabel('Loan Amount ($)')
plt.ylabel('Frequency');
```

## Observation ##

> This histogram reveals that loan amounts are unevenly distributed, with a large concentration of loans in the lower amoumt ranges. The highest frequency appears around 3,000 - 4,000 range, while additional peaks are visible around 9,000 and 15,000. The distribution is positively skewed, with fewer loans at higher amounts and a long tail extending towards 35,000. This suggests that lower-value loans are more common in the dataset than higher-value loans.

## Question 2

* What's the distribution of borrowers interest rates? 

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='BorrowerRate', bins=30)
plt.title('Distribution of Borrower Interest Rates', fontsize=14, fontweight='bold')
plt.xlabel('Borrower Interest Rate')
plt.ylabel('Frequency');
```

## Observation

> This visualization shows that borrower interest rates are concentrated mainly between approximately 10% and 25%, with the highest frequency occurring around 15% - 16%. The distribution indicates that interest rates in this range are more common among the loans in the dataset. The frequency generally declines towards the higher interest rate ranges, although some fluctuations are visible. 

## Question 3

* What's the number of loans by loan status? 

```python
plt.figure(figsize=(12, 6))
order=df['LoanStatus'].value_counts().index
sns.countplot(data=df, x='LoanStatus', order=order)
plt.title('Distribution of Loan Status', fontsize=14, fontweight='bold')
plt.xlabel('Loan Status')
plt.ylabel('Number of Loans')
plt.xticks(rotation=45, ha='right')
plt.tight_layout();
```

## Observations 

> This visualization shows that Current and Completed loans account for the largest numbers of loans in the dataset, while the other loan-status categories occur less frequently. This helps us understand the distribution of loan statuses and the repayment situations recorded on the prosper lending platform.

## Question 4

* Examine borrowers employment categories?

```python
df['EmploymentStatus'].value_counts(dropna=False)
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, y='EmploymentStatus', order=df['EmploymentStatus'].value_counts().index)
plt.title('Distribution of Borrowers by Employment Status', fontsize=14, fontweight='bold')
plt.xlabel('Number of Borrowers')
plt.ylabel('Employment Status');
```

## Observation

> This visualization shows that Employed borrowers dominate the dataset with approximately 70,000 borrowers, followed by Full-time borrowers with approximately 27,000 borrowers. While Retired and Not employed borrowers represent some of the smallest employment categories. This suggests that majority of borrowers recorded in the dataset are employed.

## Question 7

* Examine the relationship between income and loan amount?

```python
plt.figure(figsize=(12, 7))
sns.boxplot(data=df, x='IncomeRange', y='LoanOriginalAmount')
plt.title('Loan Amount by Borrowers Income Range', fontsize=14, fontweight='bold')
plt.xlabel('Income Range')
plt.ylabel('Original Loan Amount ($)')
plt.xticks(rotation=45);
```

## Observation

> The boxplot shows that borrowers earning 100,000 or more have the highest median loan amount, while borrowers earning 1-24,999 have a lower median loan amount. Borrowers in the 25,000-49,999 and 50,000-74,999 income ranges also received varying loan amounts. This suggests that borrowers in higher income ranges generally tend to receive larger loans.

## Question 8

* Compare loan amounts across employment categories?

```python
plt.figure(figsize=(12, 7))
sns.boxplot(data=df, x='EmploymentStatus', y='LoanOriginalAmount')
plt.title('Loan Amount by Employment Status', fontsize=14, fontweight='bold')
plt.xlabel('Employment Status')
plt.ylabel('Original Loan Amount ($)')
plt.xticks(rotation=45);
```

## Observation

> The box plot shows that Employed borrowers have the highest median loan amount, followed by Self-employed borrowers. Part-time and Not employed borrowers have lower median loan amounts. The chart also shows several outliers, indicating that some borrowers received unusual large loans. 

## Question 9

* What's the relationship between loan amount and borrowers interest rate?

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='LoanOriginalAmount', y='BorrowerRate', alpha=0.4)
plt.title('Relationship Between Loan Amount and Borrower Interest Rate', fontsize=14, fontweight='bold')
plt.xlabel('Original Loan Amount ($)')
plt.ylabel('Borrower Interest Rate');
```

## Observation

> The scatter plot shows a generally negative relationship between loan amount and borrower interest rate. Borrowers with larger loan amounts tend to have lower interest rates, while smaller loans show a wider range of interest rates.
